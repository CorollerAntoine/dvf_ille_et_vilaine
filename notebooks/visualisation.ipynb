```python
import pandas as pd
import plotly.express as px
import numpy as np

communes = pd.read_csv("../output/prix_m2_communes_35_2025.csv")
communes.head()
communes.shape
```

```python
communes_top20 = communes.nlargest(20, "prix_median")
fig = px.bar(communes_top20, x="prix_median", y="nom_commune", orientation="h", hover_data=["nb_ventes"], color="nb_ventes", range_color=(0, 200), labels = {"prix_median": "Prix médian (€/m²)", "nom_commune": "Commune", "nb_ventes": "Nombre de ventes"}, title="Top 20 des communes les plus chères d'Ille-et-Vilaine (prix médian au m² en 2025)")
fig.update_yaxes(categoryorder="total ascending", dtick=1)
fig.update_layout(height=700)
fig.show()
```

```python
#Nous nous sommes aperçus avec cette commandes que nous avions arrondis toutes les données sur le dvf_communes.
#Une rectification a bien été faite pour ne pas arrondir les données de latitude et longitude
communes.describe()
```

```python
fig = px.scatter_map(
    communes,
    lat="latitude",
    lon="longitude",
    hover_name="nom_commune",
    color="prix_median",
    size="nb_ventes",
    size_max=25,
    zoom=8,
    center={"lat": communes["latitude"].mean(), "lon": communes["longitude"].mean()},
    height=700,
    range_color=(1000, 5500),
    labels = {"prix_median": "Prix médian (€/m²)", "nb_ventes": "Nombre de ventes"}, 
    title="Prix médian au m² par commune — Ille-et-Vilaine, 2025",
)
fig.show()
```

```python
fig.write_html("../output/carte_prix_m2_35_2025.html")
```
